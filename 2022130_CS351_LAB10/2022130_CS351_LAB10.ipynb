**Objective**

To design and implement a Genetic Algorithm (GA) for optimizing solutions to a hypothetical game scenario where agents are evaluated based on a fitness function.

**Introduction**

A Genetic Algorithm is a nature-inspired optimization technique based on the principles of natural selection and evolution.

**Tools and Libraries**
Python 3.x
NumPy for mathematical and array operations.

**Procedure**

Implement the play_game function to evaluate the performance of agents based on their genes.

**Initialize Population:**
Randomly generate an initial population of agents within given bounds.

**Evaluate Fitness:**
Compute the fitness score of each agent using the game logic.

**Select Parents:**
Use Tournament Selection to select the most promising individuals for reproduction.

**Perform Crossover:**
Apply Uniform Crossover to produce offspring by combining genes from two parents.

**Apply Mutation:**
Introduce random changes to offspring genes with a specified mutation rate to maintain diversity.

**Iterate and Evolve:**
Repeat the process for a fixed number of generations, tracking the best solution.

**Output Results:**

Display the best solution and corresponding fitness score.
Results
Generation-wise Progress:
Display fitness improvements across generations.
Best Solution:
Final agent (genes) with the highest fitness score.
Discussion
Key Observations:
How fitness scores improved over generations.
Importance of mutation in avoiding local optima.
Potential Enhancements:
Implementing other selection methods (e.g., roulette wheel).
Testing with higher-dimensional problems or different fitness functions.
Conclusion
A Genetic Algorithm successfully optimized the agent's performance in the game.
Demonstrated the principles of selection, crossover, and mutation in evolving solutions.

```python
import numpy as np

# Define the game
def play_game(agent):
    """
    Simulates the game. Higher scores are better.
    The agent's performance depends on its genes.
    """
    score = agent[0] * 2 - agent[1] ** 2 + np.sin(agent[2] * np.pi)
    return score

# Generate initial population
def initialize_population(pop_size, gene_length, bounds):
    return np.random.uniform(bounds[0], bounds[1], (pop_size, gene_length))

# Evaluate fitness of the population
def evaluate_fitness(population):
    return np.array([play_game(agent) for agent in population])

# Select parents using tournament selection
def select_parents(population, fitness, num_parents):
    parents = []
    for _ in range(num_parents):
        tournament = np.random.choice(len(population), size=3, replace=False)
        best_in_tournament = tournament[np.argmax(fitness[tournament])]
        parents.append(population[best_in_tournament])
    return np.array(parents)

# Perform crossover (uniform crossover)
def crossover(parents, offspring_size):
    offspring = []
    for _ in range(offspring_size):
        parent1, parent2 = parents[np.random.choice(len(parents), size=2, replace=False)]
        child = np.array([np.random.choice([gene1, gene2]) for gene1, gene2 in zip(parent1, parent2)])
        offspring.append(child)
    return np.array(offspring)

# Perform mutation
def mutate(offspring, mutation_rate, bounds):
    for individual in offspring:
        if np.random.rand() < mutation_rate:
            mutation_gene = np.random.randint(len(individual))
            individual[mutation_gene] = np.random.uniform(bounds[0], bounds[1])
    return offspring

# Main Genetic Algorithm
def genetic_algorithm(pop_size, gene_length, bounds, num_generations, num_parents, mutation_rate):
    population = initialize_population(pop_size, gene_length, bounds)
    best_solution = None
    best_fitness = -np.inf

    for generation in range(num_generations):
        fitness = evaluate_fitness(population)

        # Track the best solution
        max_fitness_idx = np.argmax(fitness)
        if fitness[max_fitness_idx] > best_fitness:
            best_fitness = fitness[max_fitness_idx]
            best_solution = population[max_fitness_idx]

        print(f"Generation {generation + 1}: Best Fitness = {best_fitness:.4f}")

        parents = select_parents(population, fitness, num_parents)
        offspring_size = pop_size - len(parents)
        offspring = crossover(parents, offspring_size)
        offspring = mutate(offspring, mutation_rate, bounds)

        population = np.vstack((parents, offspring))

    return best_solution, best_fitness

# Parameters
pop_size = 20
gene_length = 3  # Each agent has 3 genes
bounds = (-5, 5)  # Each gene's value is between -5 and 5
num_generations = 50
num_parents = 10
mutation_rate = 0.1

# Run the Genetic Algorithm
best_solution, best_fitness = genetic_algorithm(
    pop_size=pop_size,
    gene_length=gene_length,
    bounds=bounds,
    num_generations=num_generations,
    num_parents=num_parents,
    mutation_rate=mutation_rate
)

print(f"Best solution found: {best_solution}, Fitness = {best_fitness:.4f}")
```
